```python
df['check_in_date_x'] = pd.to_datetime(df['booking_date'], format='%Y-%m-%d', dayfirst=True, errors='coerce')
df['check_in_date_x'] = df['check_in_date_x'].dt.normalize()
```

```python
df['checkout_date'] = pd.to_datetime(df['checkout_date'], format='%Y-%m-%d', dayfirst=True, errors='coerce')
df['checkout_date'] = df['checkout_date'].dt.normalize()
```

```python
df['booking_date']=pd.to_datetime(df['booking_date'],format='%Y-%m-%d',dayfirst=True,errors='coerce')
df['booking_date']=df['booking_date'].dt.normalize()
```
